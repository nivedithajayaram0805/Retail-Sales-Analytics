```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Hp\Desktop\internship project\retail_cleaned.csv")

df.head()
```

```python
country_orders = df['Country'].value_counts().head(10)

country_orders.plot(kind='bar')
plt.title("Order Volume by Country")
plt.xlabel("Country")
plt.ylabel("Orders")
plt.show()
```

```python
country_revenue = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10)

print(country_revenue)
```

```python
hour_sales = df.groupby('Hour')['Revenue'].sum()

hour_sales.plot(kind='bar')
plt.title("Peak Operational Hours")
plt.xlabel("Hour")
plt.ylabel("Revenue")
plt.show()
```

```python
channel_sales = df.groupby('Channel')['Revenue'].sum()

channel_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales by Channel")
plt.ylabel("")
plt.show()
```
