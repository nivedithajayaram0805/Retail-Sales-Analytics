```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Hp\Desktop\internship project\retail_cleaned.csv")

df.head()
```

```python
print(df.columns)
print(df.shape)
```

```python
print("Total Revenue =", df['Revenue'].sum())
```

```python
avg_order = df.groupby('Invoice')['Revenue'].sum().mean()

print("Average Revenue per Order =", avg_order)
```

```python
top_products = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
country_sales = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False)

print(country_sales)
```

```python
weekday_sales = df.groupby('Weekday')['Revenue'].sum()

print(weekday_sales)
```

```python
hour_sales = df.groupby('Hour')['Revenue'].sum().sort_values(ascending=False)

print(hour_sales)
```

```python
month_sales = df.groupby('Month')['Revenue'].sum()

print(month_sales)
```

```python
month_sales = df.groupby('Month')['Revenue'].sum()

print(month_sales)
```

```python
import numpy as np

df['Channel'] = np.random.choice(['Online', 'Offline'], size=len(df))

print(df[['Channel']].head())
```

```python
channel_sales = df.groupby('Channel')['Revenue'].sum()

print(channel_sales)
```
