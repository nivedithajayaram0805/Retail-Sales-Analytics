```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Hp\Desktop\internship project\retail_cleaned.csv")

df.head()
```

```python
print("Shape:", df.shape)
print(df.columns)
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
print("Total Revenue =", df['Revenue'].sum())
print("Average Revenue =", df['Revenue'].mean())
print("Total Orders =", df['Invoice'].nunique())
print("Total Products =", df['Description'].nunique())
```

```python
month_revenue = df.groupby('Month')['Revenue'].sum()

month_revenue.plot(kind='line')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
product_revenue = df.groupby('Description')['Revenue'].sum().sort_values(ascending=False).head(10)

product_revenue.plot(kind='bar')
plt.title("Revenue by Top Product Lines")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.show()
```
