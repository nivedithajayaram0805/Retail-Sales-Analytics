```python
print("Countries:", df['Country'].nunique())
print("Products:", df['Description'].nunique())
print("Invoices:", df['Invoice'].nunique())
print("Customers:", df['Customer ID'].nunique())
```

```python
country_orders = df['Country'].value_counts().head(10)

print(country_orders)
```

```python
top_products = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
country_revenue = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10)

print(country_revenue)
```

```python
month_revenue = df.groupby('Month')['Revenue'].sum()

print(month_revenue)
```

```python
weekday_revenue = df.groupby('Weekday')['Revenue'].sum()

print(weekday_revenue)
```

```python
hour_revenue = df.groupby('Hour')['Revenue'].sum().sort_values(ascending=False)

print(hour_revenue)
```

```python
channel_sales = df.groupby('Channel')['Revenue'].sum()

print(channel_sales)
```

```python
top_invoices = df.groupby('Invoice')['Revenue'].sum().sort_values(ascending=False).head(10)

print(top_invoices)
```
